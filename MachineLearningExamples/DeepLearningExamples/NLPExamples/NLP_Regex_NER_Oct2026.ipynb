```python
import re
```

```python
chat1 = "Please contact Customer support team on xyz_team@somesupport.com or call at 0123456789 0r (987)-765-5432"
```

```python
pattern1 = 'support'
match_obj = re.search(pattern1, chat1) # return first match
match_obj
```

```python
def findPattern(pattern, text):
    match = re.search(pattern, text)
    if match:
        return match.group()
    return 'No match found'
```

```python
findPattern(pattern1, chat1)
```

```python
match_obj2 = re.findall(pattern1, chat1) # return all the matches
match_obj2
```

Lets extract number

```python
pattern3 = r'\d{10}'
val = findPattern(pattern3, chat1)
val
```

```python
def get_pattern_match(pattern, text):
    all_matches = re.findall(pattern, text)
    if all_matches:
        return all_matches

```

```python
emailid = get_pattern_match(r'[a-zA-Z0-9_]*@[a-z]*\.[a-zA-Z]*', chat1)
emailid[0].strip()
```

```python
phone_num = get_pattern_match(r'(\d{10})|(\(\d{3}\)-\d{3}-\d{4})', chat1)
phone_num
```

```python
# Extract and print all matched phone numbers
for match in phone_num:
    # Join the matched groups to get the non-empty match string
    number = "".join(match).strip()
    if number:
        print(number)
```

Named Entity Recognition (NER) is a Natural Language Processing (NLP) technique that locates and classifies named entities mentioned in unstructured text into predefined categories (such as people, organizations, locations, dates, and monetary values).


How NER Works

NER systems typically process text in two main phases:
1. Entity Detection: Tokenizing the text (splitting it into words), analyzing grammatical roles via Part-of-Speech (POS) tagging, and spotting potential entity boundaries.
2. Entity Classification: Assigning a specific semantic category (e.g., PERSON, LOCATION, DATE) based on context and surrounding words.

```python
import spacy
```

```python
nlp_ner = spacy.load("en_core_web_sm")
nlp_ner.pipe_names
```

```python
text = "In Munich event , Alex said in Deutsch on 22nd Oct- our time is limited, so don't waste it without drinking beer"
doc = nlp_ner(text=text)

for entities in doc.ents:
    print(entities.text,"|", entities.label_ , "|", spacy.explain(entities.label_))
```

Lets list down all the entities that are listed in Spacy

```python
nlp_ner.pipe_labels['ner']
```

Ok lets correct and Set `Deutsch` as language

```python
from spacy.tokens import Span
# Find Deutsch in the span and set it as language
doc[7]

span1 = Span(doc, 7, 8, label="LANGUAGE")
doc.set_ents([span1], default="unmodified")
```
