To Download the full, official MNIST dataset (70,000 samples of 28 × 28 pixels) using scikit-learn, we should use the fetch_openml function of ScikitLearn

```python
from sklearn.datasets import fetch_openml

# Download the MNIST dataset from OpenML
mnist = fetch_openml('mnist_784', version=1)

# Extract features (X) and labels (y)
X, y = mnist.data, mnist.target

print(X.shape)  # Output: (70000, 784)
print(y.shape)  # Output: (70000,)
```
